# Day 7: CPU Performance & Load Averages

Day 7 covers system load average interpretation, real-time processor utilization states (`mpstat`), process-level CPU profiling, and process priority management.

## 1. System Load Averages (`uptime` & `/proc/loadavg`)
Inspect 1-minute, 5-minute, and 15-minute load averages to evaluate processor queue demand over time.

```python
uptime
cat /proc/loadavg
```

## 2. Processor Utilization States (`mpstat`)
Monitor active CPU utilization percentages across user space, system space, idle states, and I/O wait times.

```python
if command -v mpstat &> /dev/null; then
    mpstat -P ALL 1 3
else
    apt-get update && apt-get install -y sysstat
    mpstat -P ALL 1 3
fi
```

## 3. Process-Level CPU Profiling (`ps`)
Pinpoint exact processes and PIDs consuming the highest percentage of processor resources.

```python
ps aux --sort=-%cpu | head -n 10
```

## 4. CPU Core Count Inspection (`lscpu`)
Verify physical and logical core architecture to correctly baseline load averages.

```python
lscpu | grep -E 'CPU\(s\):|Core\(s\) per socket|Socket\(s\)'
```
