## 2. Finding Dangerous SUID and SGID Files
SUID (Set User ID) and SGID (Set Group ID) binaries allow users to execute files with the permissions of the file owner or group (frequently root). Auditing these files prevents privilege escalation vulnerabilities.

```python
# Find SUID binaries across the filesystem (suppressing permission denied errors)
!find / -perm -4000 -type f 2>/dev/null | head -n 20
```
