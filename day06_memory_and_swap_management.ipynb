# Day 6: Memory & Swap Management

Day 6 covers physical RAM utilization, virtual memory statistics, process-level memory profiling (RSS vs VSZ), swap space configuration, and Out-of-Memory (OOM) killer event auditing.

## 1. Physical RAM & Overall Memory Utilization (`free` & `vmstat`)
Inspect total, used, free, and available memory alongside real-time virtual memory statistics.

```python
# Check physical RAM and swap utilization in human-readable format
free -h

# Track virtual memory statistics and paging activity
vmstat 1 3
```

## 2. Process-Level Memory Profiling (`ps`)
Pinpoint exact processes and PIDs consuming the highest percentage of system RAM, comparing Resident Set Size (RSS) against Virtual Memory Size (VSZ).

```python
# List top memory-consuming processes sorted by %MEM
ps aux --sort=-%mem | head -n 10
```

## 3. Swap Space Configuration & Usage (`swapon`)
Inspect active swap partitions or files, their assigned priorities, and kernel swappiness behavior.

```python
# View active swap devices and usage
swapon --show

# Check kernel swappiness setting (how aggressively anonymous memory is swapped out)
cat /proc/sys/vm/swappiness
```

## 4. OOM (Out-of-Memory) Killer Auditing (`dmesg`)
Inspect kernel ring buffer logs to check if the system has recently terminated processes due to memory exhaustion.

```python
# Check kernel logs for OOM killer events (requires root/sudo access in full server environments)
dmesg -T | grep -i oom || echo 'No OOM killer events found or insufficient permissions to read kernel ring buffer.'
```
